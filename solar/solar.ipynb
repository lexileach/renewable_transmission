```python

from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
import geopandas as gpd

# STEP 1: Set file paths
data_file = Path("data/solar_raw.csv")
output_folder = Path("output")
output_folder.mkdir(exist_ok=True)

# STEP 2: Import solar data CSV
solar = pd.read_csv(data_file)

print("Original number of facilities:", len(solar))
print("Original columns:", solar.columns.tolist())

# STEP 3: Keep the important columns
columns = [
    "case_id",
    "p_name",
    "p_state",
    "p_county",
    "ylat",
    "xlong",
    "p_cap_ac", #Electrical Capacity \\ best to use to compare w/ transmission
    "p_cap_dc", #solar output 
    "p_year"
]

solar = solar[columns].copy()

# STEP 4: Rename columns
solar = solar.rename(columns={
    "case_id": "facility_id",
    "p_name": "facility_name",
    "p_state": "state",
    "p_county": "county",
    "ylat": "latitude",
    "xlong": "longitude",
    "p_cap_ac": "capacity_mw_ac",
    "p_cap_dc": "capacity_mw_dc",
    "p_year": "year"
})

# STEP 5: Remove duplicate facilities
solar = solar.drop_duplicates(subset="facility_id")

# STEP 6: Convert to numeric columns
numeric_cols = [
    "latitude",
    "longitude",
    "capacity_mw_ac",
    "capacity_mw_dc",
    "year"
]

for col in numeric_cols:
    solar[col] = pd.to_numeric(
        solar[col], errors="coerce"
    )

# STEP 7: Remove NA coordinates
solar = solar.dropna(
    subset=["latitude", "longitude"]
)

solar = solar[
    solar["latitude"].between(-90, 90) &
    solar["longitude"].between(-180, 180)
].copy()

# STEP 8: Filter to North Carolina
solar["state"] = solar["state"].str.strip().str.upper()
solar = solar[solar["state"] == "NC"].copy()

# STEP 9: Add source label
solar["energy_type"] = "solar"

# STEP 10: Create geographic points
solar_geo = gpd.GeoDataFrame(
    solar,
    geometry=gpd.points_from_xy(
        solar["longitude"],
        solar["latitude"]
    ),
    crs="EPSG:4326"
)

# STEP 11: Save cleaned files
solar.to_csv(
    output_folder / "solar_nc_clean.csv",
    index=False
)

solar_geo.to_file(
    output_folder / "solar_nc.geojson",
    driver="GeoJSON"
)

print("Cleaned NC solar facilities:", len(solar))
print("Files saved successfully!")
```

```python

solar_us = solar_raw.rename(columns={
    "case_id": "facility_id",
    "p_name": "facility_name",
    "p_state": "state",
    "p_county": "county",
    "ylat": "latitude",
    "xlong": "longitude",
    "p_cap_ac": "capacity_mw_ac",
    "p_cap_dc": "capacity_mw_dc",
    "p_year": "year"
}).copy()

# Keep important columns
columns = [
    "facility_id",
    "facility_name",
    "state",
    "county",
    "latitude",
    "longitude",
    "capacity_mw_ac",
    "capacity_mw_dc",
    "year"
]

solar_us = solar_us[columns].copy()

# Standardize state abbreviations
solar_us["state"] = (
    solar_us["state"].astype("string").str.strip().str.upper()
)

# Remove Alaska, Hawaii, and territories
exclude = ["AK", "HI", "PR", "VI", "GU", "MP", "AS"]

solar_us = solar_us[
    solar_us["state"].notna()
    & ~solar_us["state"].isin(exclude)
].copy()

# Convert numeric columns
numeric_cols = [
    "latitude", "longitude",
    "capacity_mw_ac", "capacity_mw_dc", "year"
]

for col in numeric_cols:
    solar_us[col] = pd.to_numeric(
        solar_us[col], errors="coerce"
    )

# Remove missing or invalid coordinates
solar_us = solar_us.dropna(
    subset=["latitude", "longitude"]
)

solar_us = solar_us[
    solar_us["latitude"].between(-90, 90)
    & solar_us["longitude"].between(-180, 180)
].copy()

# Remove duplicate facility IDs
solar_us = solar_us.drop_duplicates(
    subset="facility_id"
)

# Add energy source identifier
solar_us["energy_type"] = "solar"

# Check results
print("Facilities in lower 48 + DC:", len(solar_us))
print("States/jurisdictions:", sorted(solar_us["state"].unique()))
```

```python

import rasterio
from rasterio.windows import from_bounds
from rasterio.enums import Resampling
from rasterio.transform import from_bounds as transform_from_bounds
from rasterio.warp import transform_bounds

input_file = input_file = (
    "data/World_GHI_GISdata_LTAy_AvgDailyTotals_GlobalSolarAtlas-v2_GEOTIFF/"
    "GHI.tif"
)
output_file = "data/solar_ghi_small.tif"

with rasterio.open(input_file) as src:

    # Approximate contiguous US bounding box
    bbox = (-125, 24, -66, 50)

    # Convert bounding box to raster CRS
    bounds = transform_bounds(
        "EPSG:4326", src.crs, *bbox
    )

    window = from_bounds(
        *bounds, transform=src.transform
    )

    # Clamp window to source raster
    window = window.round_offsets().round_lengths()
    window = window.intersection(
        rasterio.windows.Window(
            0, 0, src.width, src.height
        )
    )

    # Reduce resolution by factor of 5
    new_width = max(1, int(window.width / 5))
    new_height = max(1, int(window.height / 5))

    data = src.read(
        1,
        window=window,
        out_shape=(new_height, new_width),
        resampling=Resampling.average,
        masked=True
    )

    # Calculate new transform
    new_transform = src.window_transform(window)
    new_transform = new_transform * (
        rasterio.Affine.scale(
            window.width / new_width,
            window.height / new_height
        )
    )

    profile = src.profile.copy()
    profile.update(
        height=new_height,
        width=new_width,
        count=1,
        transform=new_transform,
        compress="deflate",
        tiled=True
    )

    with rasterio.open(output_file, "w", **profile) as dst:
        dst.write(data.filled(src.nodata or -9999), 1)

print("Smaller solar raster saved!")
```

```python

import pandas as pd
import geopandas as gpd
import rasterio
import numpy as np
import matplotlib.pyplot as plt
from rasterio.enums import Resampling
from pathlib import Path

# 1. Load the cleaned nationwide solar facilities
solar = pd.read_csv("solar_lower48_clean.csv")

# Clean coordinate values
solar["latitude"] = pd.to_numeric(
    solar["latitude"], errors="coerce"
)
solar["longitude"] = pd.to_numeric(
    solar["longitude"], errors="coerce"
)

solar = solar.dropna(subset=["latitude", "longitude"])

# Convert to geographic points
solar_points = gpd.GeoDataFrame(
    solar,
    geometry=gpd.points_from_xy(
        solar["longitude"],
        solar["latitude"]
    ),
    crs="EPSG:4326"
)

# 2. Load the solar irradiance raster
raster_path = Path("data/solar_ghi_lower48.tif")

if not raster_path.exists():
    raster_path = Path("solar_ghi_lower48.tif")

with rasterio.open(raster_path) as src:
    scale = max(src.width / 1600, src.height / 1000, 1)

    ghi = src.read(
        1,
        out_shape=(
            int(src.height / scale),
            int(src.width / scale)
        ),
        resampling=Resampling.average,
        masked=True
    )

    bounds = src.bounds
    raster_crs = src.crs

ghi = ghi.astype("float32").filled(np.nan)

# Match facility coordinates to raster
solar_points = solar_points.to_crs(raster_crs)

# 3. Load state boundaries
states_url = (
    "https://www2.census.gov/geo/tiger/GENZ2023/shp/"
    "cb_2023_us_state_20m.zip"
)

states = gpd.read_file(states_url)
states = states.to_crs(raster_crs)

exclude = [
    "Alaska", "Hawaii", "Puerto Rico",
    "Guam", "American Samoa",
    "United States Virgin Islands",
    "Commonwealth of the Northern Mariana Islands"
]

states = states[~states["NAME"].isin(exclude)]

# 4. Create map
fig, ax = plt.subplots(figsize=(16, 9))

img = ax.imshow(
    ghi,
    extent=[
        bounds.left, bounds.right,
        bounds.bottom, bounds.top
    ],
    origin="upper",
    cmap="inferno"
)

# State boundaries
states.boundary.plot(
    ax=ax,
    edgecolor="white",
    linewidth=0.6
)

# Existing solar facilities
solar_points.plot(
    ax=ax,
    color="black",
    markersize=3,
    alpha=0.65,
    label="Existing solar facilities"
)

# Colorbar
cbar = plt.colorbar(img, ax=ax, shrink=0.75)
cbar.set_label("Global Horizontal Irradiation (kWh/m²/day)")

ax.legend(loc="lower left")

ax.set_title(
    "Solar Irradiance and Existing Solar Facilities "
    "Across the Contiguous United States",
    fontsize=15
)

ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")

ax.set_xlim(-125, -66)
ax.set_ylim(24, 50)

plt.tight_layout()

plt.savefig(
    "solar_irradiance_with_facilities.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()

print("Facilities with valid coordinates:", len(solar_points))

```
